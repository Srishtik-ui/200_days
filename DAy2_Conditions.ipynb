```python
#Print numbers from 1 to 15 but stop when the number becomes 10 (use break).

for i in  range(1,16):
  if i == 11:
    break
  print(i)
```

```python
#Print numbers from 1 to 10 but skip multiples of 3 (use continue).

for i in range(1,11):
  if i % 3 == 0:
    continue
  print(i)
```

```python
#Check whether a number is prime using for loop and break

num = int(input("Enter a number: "))

if num <= 1:
    print("Not Prime")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
""" Print this pattern using nested loops:
1
22
333
4444
55555"""

for i in range(1,6):
  for j in range(i):
    print(i, end="")
  print()
```
